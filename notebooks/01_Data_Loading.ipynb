# Phase 01 - Data Loading

## Objectives
- Import required libraries
- Load training and testing datasets
- Understand dataset dimensions
- Identify target column
- Verify data integrity

```python
# ==========================================================
# PHASE 01 | IMPORT LIBRARIES
# ==========================================================

import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")
```

```python
# ==========================================================
# PHASE 01 | PROJECT PATHS
# ==========================================================

PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data"
RAW_DATA_DIR = DATA_DIR / "raw"

print(f"Project Root : {PROJECT_ROOT}")
print(f"Raw Data     : {RAW_DATA_DIR}")
```

```python
# ==========================================================
# PHASE 01 | LOAD DATA
# ==========================================================

TRAIN_PATH = RAW_DATA_DIR / "train.csv"
TEST_PATH = RAW_DATA_DIR / "test.csv"
SAMPLE_SUBMISSION_PATH = RAW_DATA_DIR / "sample_submission.csv"

train_df = pd.read_csv(TRAIN_PATH)
test_df = pd.read_csv(TEST_PATH)
sample_submission = pd.read_csv(SAMPLE_SUBMISSION_PATH)

print("Datasets loaded successfully.")
```

```python
# ==========================================================
# PHASE 01 | DATASET SHAPES
# ==========================================================

print(f"Training Dataset Shape : {train_df.shape}")
print(f"Testing Dataset Shape  : {test_df.shape}")
print(f"Sample Submission Shape: {sample_submission.shape}")
```

```python
# ==========================================================
# PHASE 01 | PREVIEW DATA
# ==========================================================

print("Training Dataset")
display(train_df.head())

print("\nTesting Dataset")
display(test_df.head())
```

```python
# ==========================================================
# PHASE 01 | DATASET INFORMATION
# ==========================================================

print("Training Dataset Information")
train_df.info()

print("\n" + "=" * 60 + "\n")

print("Testing Dataset Information")
test_df.info()
```

```python
# ==========================================================
# PHASE 01 | DATASET SUMMARY
# ==========================================================

print("=" * 60)
print("TRAIN DATASET SUMMARY")
print("=" * 60)

print(f"Rows              : {train_df.shape[0]:,}")
print(f"Columns           : {train_df.shape[1]}")
print(f"Numerical Columns : {train_df.select_dtypes(include='number').shape[1]}")
print(f"Categorical Columns: {train_df.select_dtypes(include='object').shape[1]}")
print(f"Target Column     : addicted_label")
print(f"Memory Usage      : {train_df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

print("\n")

print("=" * 60)
print("TEST DATASET SUMMARY")
print("=" * 60)

print(f"Rows              : {test_df.shape[0]:,}")
print(f"Columns           : {test_df.shape[1]}")
print(f"Numerical Columns : {test_df.select_dtypes(include='number').shape[1]}")
print(f"Categorical Columns: {test_df.select_dtypes(include='object').shape[1]}")
print(f"Memory Usage      : {test_df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```
